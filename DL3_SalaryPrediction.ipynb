```python
import pandas as pd

data = pd.read_csv("multiple_linear_regression_dataset.csv")

print(data.head())
print(data.columns)
print(data.shape)

# Think about this:
# Inputs: YearsExperience, TestScore
# Output: Salary
# Number of features = 2
```

```python
# Inputs (features)
X = data[["age", "experience"]]

# Output (target)
y = data["income"]

# Shape of X: (number_of_samples, 2)
# Shape of y: (number_of_samples,)
# X has 2 columns because we have two input features,
# y has only one column because Salary is a single value to predict.
```

```python
import numpy as np

# Number of features
n_features = X.shape[1]

# Initialize weights and bias
w = np.zeros(n_features)
b = 0.0

# One weight per feature because each input contributes separately.
# Bias shifts the prediction line up or down.
# Large initial values can cause unstable learning.
```

```python
def predict(X, w, b):
    """
    X : input features
    w : weights
    b : bias
    returns predicted values
    """
    y_hat = X.dot(w) + b
    return y_hat

# No activation function because this is regression.
# y_hat can take any real value.
# Logistic regression uses an activation function and predicts probabilities.
```

```python
def mean_squared_error(y, y_hat):
    """
    y     : actual values
    y_hat : predicted values
    """
    loss = ((y_hat - y) ** 2).mean()
    return loss

# Squaring penalizes large errors more.
# One very wrong prediction increases the loss significantly.
# Absolute error is less sensitive to large mistakes.
```

```python
def compute_gradients(X, y, y_hat):
    """
    Compute gradients of MSE loss
    """
    N = len(y)

    # Gradient w.r.t weights
    dw = (2 / N) * X.T.dot(y_hat - y)

    # Gradient w.r.t bias
    db = (2 / N) * (y_hat - y).sum()

    return dw, db

# X appears in dw because weights depend on input values.
# Bias has no input, so X is not involved.
# If error is zero, gradients become zero and learning stops.
```

```python
def update_parameters(w, b, dw, db, lr):
    w = w - lr * dw
    b = b - lr * db
    return w, b
```

```python
lr = 0.0001
epochs = 1000

for epoch in range(epochs):
    # Forward pass
    y_hat = predict(X, w, b)

    # Loss calculation
    loss = mean_squared_error(y, y_hat)

    # Gradient computation
    dw, db = compute_gradients(X, y, y_hat)

    # Parameter update
    w, b = update_parameters(w, b, dw, db, lr)

    if epoch % 100 == 0:
        print(f"Epoch {epoch}, Loss: {loss}")

# Loss should decrease over time.
# If loss increases, learning rate may be too high.
# More epochs allow better convergence.
```

```python
print("Final weights:", w)
print("Final bias:", b)

# Predict salary for a new candidate
new_candidate = np.array([4.5, 68])
predicted_salary = new_candidate.dot(w) + b

print("Predicted Salary:", predicted_salary)

# Think about this:
# Prediction should be reasonable and smooth.
# Model generalizes better than fixed threshold rules.
```
