```python
import numpy as np

# STEP 1 — CREATE DATASET

# X represents the single input feature.

X = np.linspace(0, 10, 50).reshape(-1, 1)

# noise is added so the model does not memorize perfectly.

np.random.seed(42)
noise = np.random.normal(0, 0.2, size=(50, 1))
y = np.log(X + 1) + noise
print("X shape:", X.shape)
print("y shape:", y.shape)
print()


# STEP 2 — MODEL SHAPE DECISION

# We choose 3 hidden units.
# Why more than 1?
# One hidden unit behaves almost like a slightly modified linear model.
# Multiple units allow different regions of behavior.

# Why not too many?
# Too many units increase instability and overfitting risk. Data is simple, 3 is sufficient to bend smoothly.
hidden_units = 3

print("Hidden Units:", hidden_units)
print()
```

```python
# STEP 3 — PARAMETER INITIALIZATION

W1 = np.random.uniform(-1, 1, size=(1, 3))

b1 = np.zeros((1, 3))

W2 = np.random.uniform(-1, 1, size=(3, 1))

b2 = np.zeros((1, 1))
print(" PARAMETER INITIALIZATION ")
print("W1 shape:", W1.shape)
print("b1 shape:", b1.shape)
print("W2 shape:", W2.shape)
print("b2 shape:", b2.shape)
print()
```

```python
# STEP 4 — ACTIVATION FUNCTION

# ReLU.
# It keeps +ve values and suppresses negative ones.
# This introduces non-linearity.

def activation(z):
    return np.maximum(0, z)

# Slope of ReLU.
# It is:
# 0 when z <= 0
# 1 when z > 0

# Slope matters because learning depends on direction.
# Without slope, parameters would not know how to change.

def activation_slope(z):
    return (z > 0).astype(float)
```

```python
# STEP 9 — LEARNING RATE

learning_rate = 0.01

epochs = 1000
```

```python
# STEP 10 — TRAINING LOOP

for epoch in range(epochs):

    z1 = X @ W1 + b1

    h = activation(z1)

    y_hat = h @ W2 + b2

    # Error = prediction - actual
    error = y_hat - y

    loss = np.mean(error ** 2)

    dL_dy = 2 * error / len(X)

    dL_dW2 = h.T @ dL_dy
    dL_db2 = np.sum(dL_dy, axis=0, keepdims=True)

    dL_dh = dL_dy @ W2.T
    dL_dz1 = dL_dh * activation_slope(z1)

    dL_dW1 = X.T @ dL_dz1
    dL_db1 = np.sum(dL_dz1, axis=0, keepdims=True)

    #  Parameter Update


    W1 -= learning_rate * dL_dW1
    b1 -= learning_rate * dL_db1
    W2 -= learning_rate * dL_dW2
    b2 -= learning_rate * dL_db2

    if epoch % 100 == 0:
        print(f"Epoch {epoch:04d} | Loss: {loss:.6f}")

print()
print("TRAINING COMPLETE")
print("Final Loss:", loss)
print()


print("SAMPLE PREDICTIONS")
for i in range(5):
    print(f"X: {X[i][0]:.2f} | True: {y[i][0]:.4f} | Pred: {y_hat[i][0]:.4f}")
```
