```python
# Perceptron implementation

class Perceptron:
    def __init__(self, learning_rate=0.1, epochs=20):
        self.lr = learning_rate
        self.epochs = epochs
        self.weights = [0.0, 0.0]
        self.bias = 0.0

    def activation(self, x):
        return 1 if x >= 0 else 0

    def predict(self, x):
        z = self.weights[0]*x[0] + self.weights[1]*x[1] + self.bias
        return self.activation(z)

    def train(self, X, y):
        for _ in range(self.epochs):
            for i in range(len(X)):
                y_pred = self.predict(X[i])
                error = y[i] - y_pred

                # Weight and bias update
                self.weights[0] += self.lr * error * X[i][0]
                self.weights[1] += self.lr * error * X[i][1]
                self.bias += self.lr * error
```

```python
# Input combinations
X = [
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
]
```

```python
# AND Gate
y_and = [0, 0, 0, 1]
p = Perceptron()
p.train(X, y_and)

print("AND Gate")
print("Final Weights:", p.weights)
print("Final Bias:", p.bias)
for x in X:
    print(x, "->", p.predict(x))
```

```python
# OR Gate
y_or = [0, 1, 1, 1]
p = Perceptron()
p.train(X, y_or)

print("\nOR Gate")
print("Final Weights:", p.weights)
print("Final Bias:", p.bias)
for x in X:
    print(x, "->", p.predict(x))
```

```python
# NAND Gate
y_nand = [1, 1, 1, 0]
p = Perceptron()
p.train(X, y_nand)

print("\nNAND Gate")
print("Final Weights:", p.weights)
print("Final Bias:", p.bias)
for x in X:
    print(x, "->", p.predict(x))
```

```python
# NOR Gate
y_nor = [1, 0, 0, 0]
p = Perceptron()
p.train(X, y_nor)

print("\nNOR Gate")
print("Final Weights:", p.weights)
print("Final Bias:", p.bias)
for x in X:
    print(x, "->", p.predict(x))
```

```python
# XOR Gate
y_xor = [0, 1, 1, 0]
p = Perceptron()
p.train(X, y_xor)

print("\nXOR Gate")
print("Final Weights:", p.weights)
print("Final Bias:", p.bias)
for x in X:
    print(x, "->", p.predict(x))
```
